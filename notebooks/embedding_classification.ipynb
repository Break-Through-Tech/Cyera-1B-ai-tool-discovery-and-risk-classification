# Embeddings and semantic similarity — issue #7

Runs the same scripts as the command line. This compares a frozen sentence-transformer with the issue #6 TF-IDF baseline, using identical source-reviewed data and held-out applications. Results are weak-label development comparisons, not a fresh independent benchmark. See `docs/embedding_classification.md`.


```python
from pathlib import Path
import sys, subprocess, json
ROOT=Path.cwd()
if not (ROOT / "modeling").exists(): ROOT=ROOT.parent
assert (ROOT / "modeling/train_embeddings.py").exists(), "Open from the project root or notebooks folder"
def run(*args): subprocess.run([sys.executable,*map(str,args)],cwd=ROOT,check=True)
print(ROOT)
```

Install the pinned embedding dependencies. First model use downloads the public pretrained encoder.

```python
run("-m","pip","install","-r",ROOT / "modeling/requirements-embeddings.txt")
```

```python
MODELS=ROOT / "artifacts/notebook_embeddings"
run(ROOT / "modeling/train_embeddings.py","--out",MODELS,"--overwrite")
```

```python
report=json.loads((MODELS / "metrics.json").read_text())
config=json.loads((MODELS / "model_config.json").read_text())
print("Default binary family:",config["default_binary_family"])
print("Default category family:",config["default_category_family"])
for name,result in report["binary"].items():
    print(name,"test macro F1:",result["test"]["macro_f1"])
print("Category test accuracy:",report["category"]["test"]["metrics"]["accuracy"])
print("Same inspected holdout, four negative examples per split; not independent verification.")
```

```python
run(ROOT / "modeling/predict_embeddings.py","--models",MODELS,"--name","ExampleMeetingNotes","--description","An AI meeting assistant that transcribes conversations and produces summaries and action items.")
```
